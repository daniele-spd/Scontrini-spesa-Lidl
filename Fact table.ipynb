# Sostituzione Articolo con Id articolo

```python
Database = pd.read_csv("Database.csv")
```

```python
Articoli = pd.read_excel("Articoli unici.xlsx")
Articoli.drop(columns={"Prodotto"}, inplace=True)
```

```python
Fact_table = Database.merge(Articoli, how='left', on="Articolo")
Fact_table = Fact_table[["Id articolo", "IVA", "Prezzo", "Prezzo proposto", "Sconto", "Tipo sconto", "Indice scontrino"]]
Fact_table.to_csv("Fact table.csv", index=False)
```

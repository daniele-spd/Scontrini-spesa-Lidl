# Aggiornamento Excel

Processo:
* Leggo il database cercando nuovi articoli
* Leggo gli articoli unici
* Trovo gli articoli da aggiungere
* Aggiungo solo gli articoli nuovi al file Excel "Articoli unici"

```python
Database_aggiornato = pd.read_csv("Database.csv")
Database_aggiornato.drop(columns={"IVA", "Prezzo", "Prezzo proposto", "Indice scontrino"}, inplace=True)
Database_aggiornato = Database_aggiornato.sort_values("Articolo").drop_duplicates().reset_index(drop=True)

#Aggiungo un nuovo articolo di prova
# Database_aggiornato.loc[len(Database_aggiornato)] = "Nuovo articolo"
# Database_aggiornato.loc[len(Database_aggiornato)] = "Secondo articolo"
# Database_aggiornato.loc[len(Database_aggiornato)] = "Terzo articolo"
```

```python
articoli_unici = pd.read_excel("Articoli unici.xlsx")
articoli_da_aggiungere = [x for x in Database_aggiornato["Articolo"] if x not in articoli_unici["Articolo"].values]
```

```python
workbook = load_workbook("Articoli unici.xlsx")
worksheet = workbook.active

for indice, articolo in enumerate(articoli_da_aggiungere):
    worksheet.append([len(articoli_unici) + indice, articolo])

workbook.save("Articoli unici.xlsx")
```
